```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("pratyushpuri/financial-news-market-events-dataset-2025")

print("Path to dataset files:", path)
```

```python
import pandas as pd
```

```python
import os

# List the contents of the downloaded directory
print(os.listdir(path))
```

```python
import pandas as pd

# Construct the full path to the Excel file
xlsx_file_path = os.path.join(path, 'financial_news_events.xlsx')

# Load the Excel file into a pandas DataFrame
df = pd.read_excel(xlsx_file_path)

# Display the first few rows of the DataFrame
print(df.head())
```

```python
# Select only Headline and Sentiment (removing Sector)
df_RS = df[['Headline', 'Sentiment']]
df_RS.head()
```

```python
# Drop missing values from both Sentiment and Headline
df_RS1 = df_RS.dropna(subset=['Sentiment', 'Headline']).copy()
```

```python
print("Columns:", df_RS1.columns.tolist())
print("Shape:", df_RS1.shape)
df_RS1.head()
```

```python
# Save the filtered data to CSV with only Headline and Sentiment
df_RS1[['Headline', 'Sentiment']].to_csv('financial_news_events_filtered.csv', index=False)
print("Saved financial_news_events_filtered.csv locally with columns: Headline, Sentiment")
```

```python
from sklearn.model_selection import train_test_split

# Clean up labels to ensure consistency
df_RS1['Sentiment'] = df_RS1['Sentiment'].str.strip().str.lower()
valid_sentiments = ['positive', 'negative', 'neutral']
df_RS1 = df_RS1[df_RS1['Sentiment'].isin(valid_sentiments)]

# Ensure only Headline and Sentiment are preserved and Headline is string
df_RS1 = df_RS1[['Headline', 'Sentiment']]
df_RS1 = df_RS1[df_RS1['Headline'].astype(str).str.strip() != '']

# Split: 80% Train, 10% Val, 10% Test
train_df, temp_df = train_test_split(df_RS1, test_size=0.20, random_state=42, stratify=df_RS1['Sentiment'])
val_df, test_df = train_test_split(temp_df, test_size=0.50, random_state=42, stratify=temp_df['Sentiment'])

print(f"Train size: {len(train_df)} | Val size: {len(val_df)} | Test size: {len(test_df)}")
print("Columns in split:", train_df.columns.tolist())

# Save to local CSVs
train_df.to_csv('train.csv', index=False)
val_df.to_csv('val.csv', index=False)
test_df.to_csv('test.csv', index=False)
print("Saved train.csv, val.csv, and test.csv locally with columns ['Headline', 'Sentiment']!")
```
