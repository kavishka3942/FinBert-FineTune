# Upload Local Datasets to S3
This notebook reads your AWS credentials from the `.env` file and uploads your local CSV datasets to your S3 bucket. Ensure you have updated the `S3_BUCKET_NAME` in your `.env` file before running this.

```python
import os
import boto3
from dotenv import load_dotenv

# Load environment variables from .env file
load_dotenv()

AWS_ACCESS_KEY_ID = os.getenv("AWS_ACCESS_KEY_ID")
AWS_SECRET_ACCESS_KEY = os.getenv("AWS_SECRET_ACCESS_KEY")
AWS_DEFAULT_REGION = os.getenv("AWS_DEFAULT_REGION")
S3_BUCKET_NAME = os.getenv("S3_BUCKET_NAME")

if not S3_BUCKET_NAME  == "finance-sentiment-mlflow-artifacts-kavishka":
    raise ValueError("Please set a valid S3_BUCKET_NAME in your .env file.")

# Initialize S3 Client
s3_client = boto3.client(
    's3',
    aws_access_key_id=AWS_ACCESS_KEY_ID,
    aws_secret_access_key=AWS_SECRET_ACCESS_KEY,
    region_name=AWS_DEFAULT_REGION
)

print(f"Connected to S3. Ready to upload to bucket: {S3_BUCKET_NAME}")
```

```python
files_to_upload = [
    'financial_news_events_filtered.csv',
    'train.csv',
    'val.csv',
    'test.csv'
]

for file_name in files_to_upload:
    if os.path.exists(file_name):
        print(f"Uploading {file_name} to S3...")
        try:
            s3_client.upload_file(file_name, S3_BUCKET_NAME, file_name)
            print(f"✅ Successfully uploaded {file_name}")
        except Exception as e:
            print(f"❌ Failed to upload {file_name}. Error: {e}")
    else:
        print(f"⚠️ Warning: {file_name} not found locally, skipping.")
```
